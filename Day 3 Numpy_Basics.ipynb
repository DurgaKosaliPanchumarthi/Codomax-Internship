```python
# Day 3 - NumPy Basics

import numpy as np

# 1. Create Arrays
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])

print("1D Array:", arr1)
print("2D Array:\n", arr2)

# 2. Array Information
print("\nShape:", arr2.shape)
print("Dimensions:", arr2.ndim)
print("Data Type:", arr2.dtype)

# 3. Indexing
print("\nFirst Element:", arr1[0])
print("Last Element:", arr1[-1])

# 4. Slicing
print("First Three Elements:", arr1[:3])
print("Last Two Elements:", arr1[-2:])

# 5. Mathematical Operations
print("\nAddition:", arr1 + 10)
print("Multiplication:", arr1 * 2)
print("Square:", arr1 ** 2)

# 6. Array Calculations
print("\nSum:", np.sum(arr1))
print("Mean:", np.mean(arr1))
print("Maximum:", np.max(arr1))
print("Minimum:", np.min(arr1))

# 7. Matrix Operations
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("\nMatrix Addition:\n", a + b)
print("Matrix Multiplication:\n", np.dot(a, b))

print("\nDay 3 NumPy Basics Completed Successfully!")
```

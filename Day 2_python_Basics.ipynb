```python
# Variables

name = "Durga"
age = 20
cgpa = 8.4

print("Name:", name)
print("Age:", age)
print("CGPA:", cgpa)
```

```python
# Data Types

name = "Durga"
age = 20
height = 5.8
is_student = True

print(type(name))
print(type(age))
print(type(height))
print(type(is_student))
```

```python
# Operators

a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)
print("Floor Division:", a // b)
print("Exponent:", a ** b)
```

```python
# Comparison and Logical Operators

x = 10
y = 20

print(x == y)
print(x != y)
print(x < y)
print(x > y)

print(x < y and y > 10)
print(x > y or y > 10)
print(not(x > y))
```

```python
# For Loop

print("Numbers from 1 to 5:")

for i in range(1, 6):
    print(i)
```

```python
# Functions

def greet(name):
    print("Hello,", name)
    print("Welcome to Python!")

greet("Durga")
```

```python
# Function with Return

def add(a, b):
    return a + b

result = add(10, 20)

print("Sum =", result)
```

```python
# Day 2 Final Program – Python Basics

# Function to display student details
def student_details(name, age):
    print("\n----- Student Details -----")
    print("Name:", name)
    print("Age:", age)

# Function to check even or odd
def check_even_odd(num):
    if num % 2 == 0:
        print(num, "is Even")
    else:
        print(num, "is Odd")

# Main Program
print("===== Python Basics Final Program =====")

# Variables and Data Types
name = input("Enter your name: ")
age = int(input("Enter your age: "))

# Function Call
student_details(name, age)

# Operators
a = 20
b = 10

print("\n----- Operators -----")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Modulus:", a % b)

# For Loop
print("\nNumbers from 1 to 5:")
for i in range(1, 6):
    print(i)

# While Loop
print("\nCountdown:")
count = 5
while count >= 1:
    print(count)
    count -= 1

# Even or Odd
number = int(input("\nEnter a number: "))
check_even_odd(number)

print("\nProgram Executed Successfully!")
print("Thank you for learning Python Basics.")
```
